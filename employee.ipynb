```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("employee_salary_dataset.csv")
df
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
print(df.columns)
df.info()
```

```python
#average salary by department
df.groupby("Department")["Monthly_Salary"].mean().plot(kind="bar")
plt.title("Average salary by Department")
plt.xlabel("Department")
plt.ylabel("Average monthly salary")
plt.show()
```

```python
#number of employees in each department
df["Department"].value_counts().plot(kind="bar")
plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("number of employees")
plt.show()
```

```python
#Experience vs Salary
plt.scatter(df["Experience_Years"],df["Monthly_Salary"])
plt.title("Experience vs Salary")
plt.xlabel("Experience in Years")
plt.ylabel("Monthly Salary")
plt.show()
```

```python
#salary distribution
plt.hist(df["Monthly_Salary"],edgecolor="black")
plt.title("Salary Distribution")
plt.xlabel("Monthly_Salary")
plt.ylabel("Number of employees")
plt.show()
```

```python
#salary by education level
sns.barplot(data=df,x="Education_Level",y="Monthly_Salary")
plt.title("Salary by education level")
plt.xlabel("Education")
plt.ylabel("Monthly Salary")
plt.show()
```
